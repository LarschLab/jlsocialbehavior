# This is a small notebook to analyze the F2 size measurements

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Point this to your csv file.
datfolder = 'C:/Users/crodri19/OneDrive - Université de Lausanne/LarschPostDoc/QTL_mapping_project'
filename = 'Sel1_F2_size.csv'
csv_path = f'{datfolder}/{filename}'

# Load the data.
df = pd.read_csv(csv_path)

df.head()

```

```python
# Basic boxplot for all numeric columns.
# If you want one specific column, replace this with: df.boxplot(column='your_column_name')
#sns.boxplot(data=df.select_dtypes(include='number'))

fig, axes = plt.subplots(figsize=(8, 5))

allCat = df.Tank.unique()
allCat.sort()


sns.pointplot(
    data=df,
    x='Tank',
    y='Size',
    linestyle='none',
    errorbar='sd',
    order=allCat,
    color="#1F77B4",
 )

sns.swarmplot(
    data=df,
    x='Tank',
    y='Size',
    size=5,
    alpha=.6,
    order=allCat,
    color="#1F77B4",
 )

axes.set_ylabel('Size (mm)')
axes.set_xlabel('Tank')

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
